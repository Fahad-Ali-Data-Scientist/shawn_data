```python
import os
import zipfile

def zip_subfolders(input_folder, output_folder):
    os.makedirs(output_folder, exist_ok=True)

    created_count = 0
    skipped_count = 0

    for item in os.listdir(input_folder):
        item_path = os.path.join(input_folder, item)

        if os.path.isdir(item_path):
            zip_filename = os.path.join(output_folder, f"{item}.zip")

            if os.path.exists(zip_filename):
                print(f"Skipped (already exists): {zip_filename}")
                skipped_count += 1
                continue

            with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
                for root, _, files in os.walk(item_path):
                    for file in files:
                        file_path = os.path.join(root, file)
                        arcname = os.path.relpath(file_path, start=item_path)
                        zipf.write(file_path, arcname)

            print(f"Created: {zip_filename}")
            created_count += 1

    print("\nSummary:")
    print(f"Zips created: {created_count}")
    print(f"Zips skipped: {skipped_count}")
```
